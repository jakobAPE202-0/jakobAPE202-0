```python
# ==============================================================
# Celica 1: Namestitev ReportLab in sistemskih pisav DejaVu
# ==============================================================
!pip install reportlab
!apt-get install -y fonts-dejavu-core

# ==============================================================
# Celica 2: Koda za generiranje PDF z vsemi slovenskimi šumniki
# ==============================================================
import os
from reportlab.lib.pagesizes import A4
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib import colors
from reportlab.pdfgen import canvas
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont

# Registracija lokalnih sistemskih pisav DejaVu (brez spletnega prenosa)
font_reg_path = "/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf"
font_bold_path = "/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf"

pdfmetrics.registerFont(TTFont('DejaVuSans', font_reg_path))
pdfmetrics.registerFont(TTFont('DejaVuSans-Bold', font_bold_path))

class NumberedCanvas(canvas.Canvas):
    """Custom canvas za ločilno črto in noge"""
    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self._saved_page_states = []

    def showPage(self):
        self._saved_page_states.append(dict(self.__dict__))
        self._startPage()

    def save(self):
        num_pages = len(self._saved_page_states)
        for state in self._saved_page_states:
            self.__dict__.update(state)
            self.draw_page_decorations(num_pages)
            super().showPage()
        super().save()

    def draw_page_decorations(self, page_count):
        self.saveState()
        width, height = A4

        # Spodnja ločilna črta
        y_line = 45
        self.setStrokeColor(colors.HexColor("#cbd5e1"))
        self.setLineWidth(0.75)
        self.line(40, y_line, width - 40, y_line)

        # Besedilo NAD črto
        self.setFont("DejaVuSans", 9)
        self.setFillColor(colors.HexColor("#1e293b"))
        self.drawString(40, y_line + 6, "Odgovor ZZZS – Goran Mitrović")
        page_str = f"Stran {self._pageNumber} od {page_count}"
        self.drawRightString(width - 40, y_line + 6, page_str)

        # Besedilo POD črto
        self.setFont("DejaVuSans", 8)
        self.setFillColor(colors.HexColor("#64748b"))
        self.drawString(40, y_line - 14, "Carbon copy – Zdravstvo Terme Zreče")

        self.restoreState()

def build_pdf(filename="odgovor_zzzs.pdf"):
    doc = SimpleDocTemplate(
        filename,
        pagesize=A4,
        leftMargin=40,
        rightMargin=40,
        topMargin=40,
        bottomMargin=65
    )

    styles = getSampleStyleSheet()

    style_text = ParagraphStyle('TextStyle', parent=styles['Normal'], fontName='DejaVuSans', fontSize=9, leading=13, textColor=colors.HexColor("#1e293b"))
    style_body = ParagraphStyle('BodyStyle', parent=styles['Normal'], fontName='DejaVuSans', fontSize=9.5, leading=14.5, textColor=colors.HexColor("#0f172a"), spaceAfter=10)
    style_bullet = ParagraphStyle('BulletStyle', parent=styles['Normal'], fontName='DejaVuSans', fontSize=9.5, leading=14.5, textColor=colors.HexColor("#0f172a"), leftIndent=15, spaceAfter=4)
    style_salutation = ParagraphStyle('Salutation', parent=styles['Normal'], fontName='DejaVuSans-Bold', fontSize=10, leading=14, textColor=colors.HexColor("#0f172a"), spaceAfter=12)
    style_banner = ParagraphStyle('BannerStyle', parent=styles['Normal'], fontName='DejaVuSans-Bold', fontSize=10, leading=14, textColor=colors.white)

    story = []

    # --- GLAVA DOKUMENTA ---
    prejemnik_html = """
    <b>Prejemnik (Recipient):</b><br/>
    Zavod za zdravstveno zavarovanje Slovenije (ZZZS)<br/>
    Območna enota Ljubljana<br/>
    Miklošičeva cesta 24, 1000 Ljubljana
    """

    izvajalec_html = """
    <b>Izvajalec storitve (V vednost):</b><br/>
    Unitur d.o.o., Terme Zreče – Recepcija zdravstva<br/>
    Cesta na Roglo 15, 3214 Zreče
    """

    posiljatelj_html = """
    <b>Pošiljatelj (Sender):</b> Jakob Seliškar<br/>
    <b>KZZ št.:</b> 035718722<br/>
    <b>Št. delovnega naloga:</b> 2026082705547
    """

    # Prejemnik (levo) + prostor desno zgoraj za e-podpis
    t1_data = [
        [Paragraph(prejemnik_html, style_text), Paragraph("", style_text)]
    ]
    t1 = Table(t1_data, colWidths=[280, 235])
    t1.setStyle(TableStyle([
        ('VALIGN', (0,0), (-1,-1), 'TOP'),
        ('LEFTPADDING', (0,0), (-1,-1), 0),
        ('RIGHTPADDING', (0,0), (-1,-1), 0),
        ('BOTTOMPADDING', (0,0), (-1,-1), 0),
        ('TOPPADDING', (0,0), (-1,-1), 0),
    ]))
    story.append(t1)

    story.append(Spacer(1, 15))

    # Izvajalec (levo) + Pošiljatelj (desno, poravnano)
    t2_data = [
        [Paragraph(izvajalec_html, style_text), Paragraph(posiljatelj_html, style_text)]
    ]
    t2 = Table(t2_data, colWidths=[280, 235])
    t2.setStyle(TableStyle([
        ('VALIGN', (0,0), (-1,-1), 'TOP'),
        ('LEFTPADDING', (0,0), (-1,-1), 0),
        ('RIGHTPADDING', (0,0), (-1,-1), 0),
        ('BOTTOMPADDING', (0,0), (-1,-1), 0),
        ('TOPPADDING', (0,0), (-1,-1), 0),
    ]))
    story.append(t2)

    story.append(Spacer(1, 14))

    story.append(Paragraph("<b>Kraj in datum:</b> Zreče, 08.11.2026", style_text))
    story.append(Spacer(1, 12))

    # --- ZELENA VRSTICA (BANNER) ---
    banner_text = Paragraph("Re: Ohranitev odprtega delovnega naloga – Seliškar Jakob 27569/26", style_banner)
    banner_table = Table([[banner_text]], colWidths=[515])
    banner_table.setStyle(TableStyle([
        ('BACKGROUND', (0,0), (-1,-1), colors.HexColor("#1b8047")),
        ('TOPPADDING', (0,0), (-1,-1), 6),
        ('BOTTOMPADDING', (0,0), (-1,-1), 6),
        ('LEFTPADDING', (0,0), (-1,-1), 10),
        ('RIGHTPADDING', (0,0), (-1,-1), 10),
        ('VALIGN', (0,0), (-1,-1), 'MIDDLE'),
    ]))
    story.append(banner_table)

    story.append(Spacer(1, 16))

    # --- VSEBINA ---
    story.append(Paragraph("Spoštovani g. Goran Mitrović, dr. med., spec.,", style_salutation))
    story.append(Spacer(1, 6))

    story.append(Paragraph("Zahvaljujem se vam za posredovano pojasnilo glede pristojnosti imenovanega zdravnika.", style_body))
    story.append(Paragraph("Pojasnjujem, da se ambulantna fizioterapevtska obravnava na delovni nalog v Termah Zreče (Unitur d.o.o.) na podlagi strokovnega dogovora z izvajalcem izvaja v dveh delih:", style_body))

    story.append(Paragraph("• <b>Prvi del (10 obravnav):</b> od 03.11. do 13.11.", style_bullet))
    story.append(Paragraph("• <b>Drugi del (5 obravnav):</b> v naknadno dogovorjenem terminu z izvajalcem.", style_bullet))

    story.append(Spacer(1, 6))

    story.append(Paragraph("Zaradi razdalje med mestom mojega bivanja (Ljubljana) in krajem izvajanja obravnave (Zreče) sta dinamika obiskov ter način prihodov oziroma samoplačniške nastanitve izključno moja lastna odločitev in organizacija.", style_body))
    story.append(Paragraph("Ker imam v naslednjih dveh tednih objektivne razloge, zaradi katerih ne morem vseh 15 obravnav fizioterapije zaključiti v neprekinjenem bivanju v Zrečah, dnevno pa se iz Ljubljane ne bom vozil, se bom za izvedbo preostalih 5 obravnav v sklopu terminov za vsak ambulantni obisk dogovarjal neposredno z izvajalcem Zdravstvo Terme Zreče (Unitur d.o.o.), kjer imam ureditve glede bivanja – česar Zakon o zdravstvenem varstvu in zdravstvenem zavarovanju (ZZVZZ) ne prepoveduje.", style_body))
    story.append(Paragraph("Poudarjam, da s tem ne uveljavljam pravice do stacionarnega zdraviliškega zdravljenja ali bivanja na račun obveznega zdravstvenega zavarovanja (ZZZS), temveč gre izključno za logistično prilagoditev izvajanja odobrenega ambulantnega delovnega naloga. Vsakodnevna vožnja na relaciji Ljubljana–Zreče bi namreč zaradi zdravstvenega stanja ramena/roke predstavljala nesrazmerno telesno obremenitev ter medicinsko nesmotrnost, ki bi neposredno zmanjšala učinek izvedenih terapij.", style_body))

    story.append(Spacer(1, 10))

    # --- PODPIS ---
    story.append(Paragraph("Lepo pozdravljeni,", style_body))
    story.append(Paragraph("<b>Jakob Seliškar</b>", style_body))

    doc.build(story, canvasmaker=NumberedCanvas)
    print(f"Dokument '{filename}' je bil uspešno ustvarjen z vsemi slovenskimi šumniki!")

if __name__ == "__main__":
    build_pdf()
```
